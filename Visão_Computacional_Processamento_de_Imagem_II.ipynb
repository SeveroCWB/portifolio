# * ETAPA 1: Carregue a biblioteca OpenCV*

```python
import cv2
import urllib.request
import numpy as np
from google.colab.patches import cv2_imshow

print("=== EXERCÍCIO DE SEGMENTAÇÃO - ETAPAS DO PROCESSAMENTO ===\n")
```

# * ETAPA 2: Importe uma imagem colorida da web*

```python
print("ETAPA 1: Carregando bibliotecas... ✓")
print("ETAPA 2: Importando imagem colorida da web...")

# Usando a imagem do link fornecido
url_imagem = "https://www.palpitedigital.com/y/5327/imagens-google-e1604596848141-599.webp"

# Baixar e carregar a imagem
try:
    resp = urllib.request.urlopen(url_imagem)
    imagem_array = np.asarray(bytearray(resp.read()), dtype=np.uint8)
    imagem_colorida = cv2.imdecode(imagem_array, cv2.IMREAD_COLOR)
    imagem_colorida = cv2.cvtColor(imagem_colorida, cv2.COLOR_BGR2RGB)

    print("✅ Imagem colorida carregada com sucesso!")
    print(f"Dimensões: {imagem_colorida.shape}")

    # Mostrar imagem original colorida
    print("\n📷 Imagem Original Colorida:")
    cv2_imshow(imagem_colorida)

except Exception as e:
    print(f"❌ Erro ao carregar a imagem: {e}")
    print("Verifique se a URL está correta e acessível.")
```

### ETAPA 3: Converta a imagem para escala de cinza

```python
print("\nETAPA 3: Convertendo para escala de cinza...")
imagem_cinza = cv2.cvtColor(imagem_colorida, cv2.COLOR_RGB2GRAY)
```

# *ETAPA 4: Mostre a imagem*

```python
print("ETAPA 4: Mostrando imagem em escala de cinza:")
cv2_imshow(imagem_cinza)
print("✓ Imagem em escala de cinza exibida")
```

# *ETAPA 5: Realize a "binarização" da imagem*

```python
print("\nETAPA 5: Realizando binarização da imagem...")
_, imagem_binarizada = cv2.threshold(imagem_cinza, 127, 255, cv2.THRESH_BINARY)
```

# *ETAPA 6: Mostre a imagem*

```python
print("ETAPA 6: Mostrando imagem binarizada:")
cv2_imshow(imagem_binarizada)
print("✓ Imagem binarizada exibida")
```

# *ETAPA 7: Faça a subtração do fundo, mantendo a parte definida de interesse*

```python
print("\nETAPA 7: Realizando subtração do fundo...")
mask = imagem_binarizada.copy()
fundo_removido = cv2.bitwise_and(imagem_colorida, imagem_colorida, mask=mask)
```

# *ETAPA 8: Mostre a imagem*

```python
print("ETAPA 8: Mostrando imagem com fundo removido:")
cv2_imshow(fundo_removido)
print("✓ Subtração de fundo concluída")
```

# *ETAPA 9: Execute a detecção das bordas da parte definida de interesse*

```python
print("\nETAPA 9: Executando detecção de bordas...")
bordas = cv2.Canny(imagem_cinza, 100, 200)
print("ETAPA 9: Detecção de bordas concluída")
print("ETAPA 10: Mostrando imagem com detecção de bordas:")
cv2_imshow(bordas)
print("✓ Imagem com bordas detectadas exibida")
```
